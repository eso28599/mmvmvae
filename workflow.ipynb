<summary>Working on a submission system</summary>
To train the classifier on the CelebA dataset for evaluation of VAE outputs, run:

```bash
qsub run_pm/pm_classifier.pbs
```
And then to run a specific experiment via submission system
```bash
qsub run_pm/polyMNIST_general.pbs
```
where the `model` and `model.aggregation` variables must be edited in the file `run_pm/polyMNIST_general.pbs` so that the desired model is used. 

| Model       | `model`     |`model.aggregation` |
| ----------- | ----------- |----------- |
| JPVAE      | jointprior   | avg |
| MMVMVAE   | joint        | avg |
| MMVMVAE   | mixedprior        | avg |
| MoE   | mixedprior        | moe |
| MoPoE   | mixedprior        | mopoe |
| independent   | unimodal        | avg |

To generate the parameters needed for the experiments:
```bash
./run_PM/generate_params_PM.sh 
```

and to run the experiments: check this one !!!
```bash 
qsub run_pm/run_polymnist.sh
```
This runs both the experiment for all methods using M0 as the first view for JPVAE and the experiment investigate using different views as the first modality. 

If you want to run only the former experiment, the line `#PBS -J 1-90` should be replaced with `#PBS -J 1-70`, and the line `# params=$(sed -n "${PBS_ARRAY_INDEX}p" run_PM/pm_params.txt) # main experiment` should be uncommented. Similarly, if you want to run only the latter experiment, the line `#PBS -J 1-90` should be replaced with `#PBS -J 71-90`, and the line `# params=$(sed -n "${PBS_ARRAY_INDEX}p" run_PM/pm_params_mod.txt) # mod ` should be uncommented.
