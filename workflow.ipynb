And then to run a specific experiment via submission system
```bash
qsub run_celeba/celeba_general.pbs
```
where the `model` and `model.aggregation` variables must be edited in the file `run_celeba/celeba_general.pbs` so that the desired model is used. 

| Model       | `model`     |`model.aggregation` |
| ----------- | ----------- |----------- |
| JPVAE      | jointprior   | avg |
| MMVMVAE   | joint        | avg |
| MMVMVAE   | mixedprior        | avg |
| MoE   | mixedprior        | moe |
| MoPoE   | mixedprior        | mopoe |
| independent   | unimodal        | avg |

* "jointprior": JPVAE
* "joint": MMVMVAE
* "mixedprior":  
* "unimodal": independent VAEs for each modality
